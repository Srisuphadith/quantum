```python
import cirq
```

```python
q0 = cirq.GridQubit(0,0)
q1 = cirq.GridQubit(0,1)
```

```python
m0 = cirq.X(q0)
m1 = cirq.CNOT(q0,q1)
m2 = cirq.measure(q0,q1)
```

```python
cir = cirq.Circuit()
cir.append(m0)
cir.append(m1)
cir.append(m2)
```

```python
print(cir)
```

```python
sim0 = cirq.Simulator()
result0 = sim0.run(cir,repetitions=10)
print(result0)
```

```python
sim1 = cirq.Simulator()
result1 = sim1.simulate(cir)
print(result1)
```
